```python
# Author: Cameron Liu 
# Purpose: To graph all the information from the simulations, more importantly for the cumulative mean

import matplotlib.pyplot as plt
import numpy as np
import statistics      
from scipy import stats

# Set global theme settings for plotting
plt.rcParams.update({
    'font.size': 20,          # Set font size
    'lines.linewidth': 3,     # Set default line thickness
    'axes.titlesize': 20,     # Title font size
    'axes.labelsize': 18,     # Axis label font size
    'xtick.labelsize': 14,    # X-tick label font size
    'ytick.labelsize': 14,    # Y-tick label font size
    'grid.color': 'gray',     # Grid color
    'grid.linestyle': '--',   # Grid line style
    'grid.linewidth': 0.5,    # Grid line width
    'legend.fontsize': 16     # Legend font size
})

all_datasets = [] # All of the datasets, that are stored as tuples
timesteps = [] # Timesteps, 0 to 20,000


def parse_files():
    # Purpose: Parses the files by building strings in order to gather all the data from the file
    # 
    # Returns: A list of tuples. (A, B). Let A be the number associated with the file, and B be the Data from the file. 
    for i in range(20):
        filename = "/Users/cameronliu/Desktop/Research/sd-0.0025/datafor_relative_tskitstatus_OFF_fixationcalc_OFF_Sb_1.0000_deldist_point_bendist_exponential_mub_0.0000_chromnum_23_N0_2000_mud_2_L_200_seed_24_Sd_0.025000/rawdataforNxtimesteps20000popsize2000mutrate2chromsize200chromnum23benmutrate0.0000Sb1.0000.txt"
        file_data = []
        with open(filename, "r") as file:
            next(file) # skips the first line
            for line in file: 
                line = line.strip().split(",")
                if len(timesteps) < 20000: # Skips adding the 0 - 20,000 timesteps
                    timesteps.append(line[0])
                file_data.append(float(line[3]))
        all_datasets.append((i, file_data))
    return all_datasets

def calculate_cumulative_mean(position, dataset): 
    # Purpose: calculates the cumulative mean which corresponds to the type of mean that it also needs
    # 
    # The method needs the current position that you want the cumulative mean at, and the dataset that corresponds to it.
    #
    # Returns: The current mean, up to that position, respective to which type of mean (will add later)
    if len(dataset[:position]) == len(timesteps[:position]):
        return dataset[:position] / timesteps[:position]
    



parse_files()
```
